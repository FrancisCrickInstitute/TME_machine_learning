```python
from prostate_recurrence.feature_engineering.Python import glcm
```

```python
input_directory = "/Users/fux/Documents/Projects/DNN-prostate-ecm/2021m08__feature_engineering_texture/data_TWOMBLI_deconv/colour_1/"
```

```python
image_names = os.listdir(input_directory)
print(image_names)
```

```python
image_name = '1811 A2 edge 1.tif-(Colour_1).tif'
image_name = '1811 A2 core 1.tif-(Colour_1).tif'
image_name = '1811 A2 adipose 1.tif-(Colour_1).tif'
```

```python
path_to_img = os.path.join(
    input_directory, image_name
)
```

```python
image_arr = glcm.read_image(path_to_img)
image_arr = 255-image_arr
```

```python
distances = [1,2,5]
angles = [0, np.pi/4., np.pi/2., np.pi*3/4.]
symmetric = True
normed = True
```

```python
matrix_glcm = glcm.construct_glcm(
    image=image_arr,
    distances=distances,
    angles=angles,
    symmetric=symmetric,
    normed=normed
)
```

```python
glcm_features_output = glcm.extract_glcm_features(
    matrix=matrix_glcm,
    features=(
        "contrast",
        "dissimilarity",
        "homogeneity",
        "energy",
        "correlation",
        "ASM",
    )
)
```

```python
matrix_glcm.shape
```

```python
output_directory = os.path.join(
    input_directory,
    f"{image_name.split('.')[0]} glcm outputs"
)
os.makedirs(output_directory, exist_ok=True)
```

```python
glcm.save_glcm_features(
    matrix=matrix_glcm,
    glcm_features_output=glcm_features_output,
    distances=distances,
    angles=angles,
    output_directory=output_directory
)
```
