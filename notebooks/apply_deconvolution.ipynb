```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import histomicstk as htk   # pip install histomicstk
from glob import glob
```

```python
stain_color_map = {
    'psr' : [0.174309, 0.8309804, 0.5282877],
    'nuc1' : [0.37535256, 0.61940926, 0.6852346],
    'nuc2' : [0.22552978, 0.6076955, 0.7614739]
}
stains = ['psr', 'nuc1', 'nuc2']
W = np.array(
    [stain_color_map[st] for st in stains]
).T
```

```python
image_paths = glob(
    "../../../2021m06__tiling_deconvolution/data/C032019_B04_PSR.czi_tile_size_512/*.tif"
)
```

```python
for image_path in image_paths:
    img_arr = np.array(Image.open(image_path))
    
    imDeconvolved = htk.preprocessing.color_deconvolution.color_deconvolution(img_arr, W)
    
    for channel, stain in enumerate(stains):
        save_path = os.path.join(
            os.path.dirname(image_path),
            os.path.basename(image_path).split('.')[0] + f"_{stain}.tif"
        )
        
        imDec_arr = np.dstack([imDeconvolved.Stains[:,:,channel]]*3)
        
        imDec = Image.fromarray(imDec_arr)
        
        print(f"> saving {save_path}")
        imDec.save(save_path)
```
