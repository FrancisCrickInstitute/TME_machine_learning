## stitch grey scale PSR

```python
# Change this to your local directory !
image_paths = glob(
    "../../../2021m06__tiling_deconvolution/data/C032019_B04_PSR.czi_tile_size_512/*_psr.tif"
    #"../../../2021m06__tiling_deconvolution/data/C032019_B04_PSR.czi_tile_size_512/PSR_C032019_B04_PSR_image_tile_008_019_psr.tif_coloured.tif"
)
```

```python
save_path = "../../../2021m06__tiling_deconvolution/data/C032019_B04_PSR.czi_tile_size_512/PSR_C032019_B04_whole_slide_stitched.tif"
```

```python
nrow = 1; ncol = 1
for image_path in image_paths:
    image_name = image_path.split('/')[-1]
    image_name_parts = image_name.split("_")
    position_str_tile = image_name_parts.index("tile")

    tile_row = int(image_name_parts[position_str_tile + 1])
    tile_col = int(image_name_parts[position_str_tile + 2])
    
    if tile_row + 1 > nrow:
        nrow = tile_row + 1
    if tile_col + 1 > ncol:
        ncol = tile_col + 1

print(nrow, ncol)
```

```python
stitching.reconstruct_whole_slide(
    file_paths=image_paths,
    save_path=save_path,
    nrow=nrow,
    ncol=ncol,
    dim=(512,512)
)
```
