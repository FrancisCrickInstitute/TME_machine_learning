## batch processing

```python
data_paths = natsorted(glob(
    "/Extra/Extra2/deep_learning/project_ML_prostate_recurrence/workspace_willow/data/*czi"
))
print(data_paths)
```

```python
for data_path in data_paths[1:2]:
    sample_id = os.path.basename(data_path).split('.')[0]
    print(f'> sample : {sample_id}')
    
    dict_imgs_with_res = tissue_mask.read_image(
        path_to_img=data_path,
        method='bioformats',
        highest_resolution='20x',
        resolution='5x'
    )
    plt.imshow(
        dict_imgs_with_res['5.0x'][
            'ScanRegion0'
        ]
    )
    plt.close()
    dict_masks = tissue_mask.create_tissue_mask(
        dict_imgs=dict_imgs_with_res['5.0x'],
        gaussian_blur_sigma=2.,
        resize_factor=4,
        erosion_n=10,
        dilation_n=10
    )
    plt.imshow(
        dict_masks[
            'ScanRegion0'
        ], 
        cmap=plt.cm.Greys
    )
    plt.close()
    
    save_path=f'/Extra/Extra2/deep_learning/project_ML_prostate_recurrence/workspace_willow/notebooks/whole_slide_images_low_resolution/{sample_id}/'
    os.makedirs(save_path, exist_ok=True)
    
    print('... saving low resolution whole slide')
    tissue_mask.save_low_res_whole_slide_image(
        dict_imgs=dict_imgs_with_res['5.0x'],
        dict_masks=dict_masks,
        slide_id=sample_id,
        save_path=save_path,
        resolution='5.0x'
    )
    
    
    # save mask tiles at raw resolution
#     print('... saving high resolution mask tiles')
#     tile_size = 2000
#     for img_name, mask in dict_masks.items():
#         original_shape = dict_imgs_with_res['20x'][img_name].shape
#         mask_large = tissue_mask.resize(mask, original_shape)
        
#         (dict_mask_tiles, nrow, ncol) = tissue_mask.create_mask_tiles(
#             img=mask_large,
#             size=tile_size
#         )
        
#         save_path_subdir = os.path.join(
#             save_path,
#             img_name,
#             f'tile_size_{tile_size}'
#         )
#         os.makedirs(save_path_subdir)
        
#         tissue_mask.save_mask_tiles(
#             dict_img_tiles=dict_mask_tiles,
#             nrow=nrow,
#             ncol=ncol,
#             save_path=save_path_subdir
#         )
```
